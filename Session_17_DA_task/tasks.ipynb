# Task 3

### Create a Python program that accepts a date in 'YYYY-MM-DD' format from the user and displays the day of the week using the datetime module.

```python
import datetime

date = input("Enter date (YYYY-MM-DD): ")
date = datetime.datetime.strptime(date, "%Y-%m-%d")
day = date.strftime("%A")
print("Day of the week:", day)
```

### Build a simple custom module named insta_utils.py with a function format_follower_count(n) that returns '1.5K' for 1500 and '2.3M' for 2300000. Import and use this function in another script to display formatted counts for 3 sample numbers.

Step 1: Create a file named insta_utils.py

```python
%%writefile insta_utils.py

def format_follower_count(n):
    if n >= 1000000:
        return str(round(n / 1000000, 1)) + "M"

    elif n >= 1000:
        return str(round(n / 1000, 1)) + "K"

    else:
        return str(n)
```

Step 2: Create another file named main.py

```python
import insta_utils

print(insta_utils.format_follower_count(1500))
print(insta_utils.format_follower_count(2300000))
print(insta_utils.format_follower_count(500))
```
